Fill all missing values with NA Value, except release year which is dropped three rows. release_year dropped to keep the columns data type as numeric, meanwhile the others can be filled with NA since its a string data type columns

```python

  df = df.dropna(how="any", subset=['release_year'])
  df = df.dropna(how="any", subset=['date_added'])
  df = df.fillna('NA')
```
